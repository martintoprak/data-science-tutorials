# Who are the survivors of the Titanic?

#### Survival rate

```python
df.groupby('Survived')['PassengerId'].agg(['count'])
```

```python
df.groupby('Survived')['PassengerId'].agg(
    ['count']
).reset_index().plot(x='Survived', y='count', kind = 'bar', figsize = (10, 10))
```

#### Survival rate depending on genre

```python
df.groupby(
    ['Survived', 'Sex']
)['PassengerId'].agg(['count'])
```

```python
df.groupby(
    ['Survived', 'Sex']
)['PassengerId'].agg(['count']).unstack()
```

```python
df.groupby(
    ['Survived', 'Sex']
)['PassengerId'].count().unstack().plot(kind ='bar', figsize = (10, 10))
```

#### Survival rate depending on Ticket class

```python
df.groupby(['Survived', 'Pclass'])['PassengerId'].agg(
    ['count']
)
```

```python
df.groupby(['Survived', 'Pclass'])['PassengerId'].count().unstack()
```

```python
df.groupby(
    ['Survived', 'Pclass']
)['PassengerId'].count().unstack().plot(kind ='bar', figsize = (10, 10))
```

```python

df['starboard'] = (
    df['Cabin'].str[-1].isin(['1', '3', '5', '7', '9']).astype(int)
    .where(df['Cabin'].notna())
)
print(df['starboard'].count())
```

```python
df.groupby(
    ['Survived', 'starboard']
)['PassengerId'].count().unstack().plot(kind ='bar', figsize = (10, 10), legend = '')
```

```python
import numpy as np
import matplotlib.pyplot as plt

counts = df.groupby(['Survived', 'starboard'])['PassengerId'].count().unstack()

x = np.arange(len(counts.index))   # one position per Survived value (0, 1)
width = 0.4

plt.figure(figsize=(10, 10))
plt.bar(x - width/2, counts[0], width, label='Port (even)')
plt.bar(x + width/2, counts[1], width, label='Starboard (odd)')

plt.xticks(x, ['Died', 'Survived'])
plt.ylabel('Number of passengers')
plt.legend(title='Cabin side')
plt.show()
```

```python
df['family_size'] = df['SibSp'] + df['Parch'] + 1   # +1 for the passenger themselves
df['has_family'] = (df['family_size'] > 1).astype(int)
```

```python

counts = df.groupby(['Survived', 'has_family'])['PassengerId'].count().unstack()

x = np.arange(len(counts.index))
width = 0.4

plt.figure(figsize=(10, 10))
plt.bar(x - width/2, counts[0], width, label='has no family')
plt.bar(x + width/2, counts[1], width, label='has family')

plt.xticks(x, ['Died', 'Survived'])
plt.ylabel('Number of passengers')
plt.legend(title='family')
plt.show()
```
